```python

plt.figure(figsize=(14, 6))


plt.plot(df['date'], df['price'], label='Original Price', color='black', alpha=0.7)

plt.plot(df['date'], df['price'].rolling(30).mean(), label='30-Day Moving Average', color='blue', linewidth=1.5)
plt.plot(df['date'], df['price'].rolling(90).mean(), label='90-Day Moving Average', color='orange', linewidth=1.5)

plt.yscale('log')  
plt.title('Bitcoin Price vs Moving Averages (Trend Smoothing - Log Scale)')
plt.xlabel('Date')
plt.ylabel('Price in USD (Log Scale)')
plt.legend()
plt.grid(True, which='both', linestyle='--', alpha=0.5)
plt.show()
```

```python
plt.Figure(figsize=(12,5))
sns.histplot(df['change_pct'],bins=150,kde=True,color='purple')
plt.xlim(-25,25)
plt.axvline(0, color='red', linestyle='--', linewidth=1.5, label='Zero Change (0%)')
plt.title("Bitcoin Daily Volatility & Percentage Change Distribution")
plt.xlabel("Daily Change Percentage (%)")
plt.ylabel("Frequency (Days Count)")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
plt.figure(figsize=(12,5))
sns.scatterplot(x=df['vol'],y=df['change_pct'],alpha=0.4,color='teal')

plt.title("Volume vs Daily Price Change Percentage")
plt.xlabel("Trading Volume")
plt.ylabel("Daily Change Percentage (%)")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
plt.figure(figsize=(15,5))
corr = df[['price', 'open', 'high', 'low', 'vol', 'change_pct']].corr()
sns.heatmap(corr,annot=True,cmap="coolwarm",fmt=".2f",linewidths=0.5,annot_kws={"size": 13})
plt.title("Feature Correaltion Heatmap")
plt.tight_layout()
plt.show()
```
